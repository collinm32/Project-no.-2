## 1. Function Q: Water Weight (lb/yd³)
▶️ **Code Cell:** define `calc_water_weight(...)`

**Inputs:**
- cement content
- w/c ratio
- (<any modifiers from Excel>)

**Output:**
- water weight (lb/yd³)

---


```python
def calculate_water_weight_Q(cement_A, fly_ash_B, silica_fume_C, other_SCM_D, wc_ratio_E):
    total_cementitious = cement_A + fly_ash_B + silica_fume_C + other_SCM_D
    water_weight_Q = total_cementitious * wc_ratio_E
    return water_weight_Q
# This function is used to find the water weight of the cementitious material.
```

**Inputs:** cement weight (lb), fly ash weight (lb), silica fume weight(lb), other SCM weight (lb), and water to concrete ratio

**Outputs:** water weight (lb/yd^3)

```python
def calculate_volume_R(cement_A, sg_cement_J):
    volume_R = cement_A / (sg_cement_J * unit_weight_water)
    return volume_R
# This function is used to find the volume of cement given the weight and specific gravity of the cement.
```

**Inputs:** cement weight (lb), specific gravity of cement, and unit weight of water (lb/ft^3)

**Outputs:** volume of cement (ft^3)

```python
# Define a function to calculate fly ash volume
def calculate_volume_S(fly_ash_B, sg_fly_ash_K):
    volume_S = fly_ash_B / (sg_fly_ash_K * unit_weight_water)
    return volume_S
# This function is used to find the volume of fly ash given the weight and specific gravity of the fly ash.
```

**Inputs:** fly ash weight (lb), specific gravity of fly ash, and unit weight of water (lb/ft^3)

**Outputs:** volume of fly ash (ft^3)

```python
# Define a function to calculate silica fume volume
def calculate_volume_T(silica_fume_C, sg_silica_fume_L):
    volume_T = silica_fume_C / (sg_silica_fume_L * unit_weight_water)
    return volume_T
# This function is used to find the volume of silica fume given the weight and specific gravity of the silica fume.
```

**Inputs:** weight of silica fume (lb), specific gravity of silica fume, and the unit weight of water (lb/ft^3)

**Outputs:** volume of silica fume (ft^3)

```python
def calculate_volume_U(other_SCM_D, sg_SCM_M):
    volume_U = other_SCM_D / (sg_other_SCM_M * unit_weight_water)
    return volume_U
# This function is used to find the volume of Other SCM given the weight and specific gravity of the Other SCM.
```

**Inputs:** other SCM weight (lb), specific gravity of other SCM, and unit weight of water (lb/ft^3)

**Outputs:** volume of other SCM (ft^3)

```python
def calculate_volume_V(air_content_F):
    volume_V = (air_content_F/100) * cubic_yard_ft3
    return volume_V
# This function is used to find the volume of air given the target air content and the cubic feet to cubic yard conversion.
```

**Inputs:** target air content (%) and cubic feet into one cubic yard conversion

**Outputs:** volume of air (ft^3)

```python
def calculate_volume_W(water_weight_Q):
    return water_weight_Q / unit_weight_water
# This function is used to find the volume of water given the weight of water and the unit weight of water.
```

**Inputs:** weight of water (lb) and unit weight of water (lb/ft^3)

**Outputs:** volume of water (ft^3)

```python
def calculate_total_aggregate_volume_X(volume_R, volume_S, volume_T, volume_U, volume_V, volume_W):
    volume_X = cubic_yard_ft3 - volume_R - volume_S - volume_T - volume_U - volume_V - volume_W
    return volume_X
# This function is used to find the volume of total aggregates using all the volumes previously calculated and the cubic feet to cubic yard conversion.
```

**Inputs:** volume of cement (ft^3), volume of fly ash (ft^3), volume of silica fume (ft^3), volume of air (ft^3), volume of water (ft^3), and cubic feet into one cubic yard conversion

**Outputs:** volume of total aggregates (ft^3)

```python
def calculate_fine_aggregate_Y(percent_fine_G, sg_fine_N, volume_X):
    weight_Y = unit_weight_water * (percent_fine_G / 100) * sg_fine_N * volume_X
    return weight_Y
# This function is used to find the total weight of the fine aggregates, the percent fine aggregate that you want, the specific gravity of fine aggregate, the volume of total aggregate, and the unit weight of water.
```

**Inputs:** percent of fine aggregate (%), specific gravity of fine aggregate, volume of total aggregate (ft^3), and unit weight of water (lb/yd^3)

**Outputs:** total weight of fine aggregate (lb)

```python
def calculate_coarse_aggregate_Z(percent_coarse_H, sg_coarse_O, volume_X):
    weight_Z = unit_weight_water * (percent_coarse_H / 100) * sg_coarse_O * volume_X
    return weight_Z
# This function is used to find the total weight of the coarse aggregates, the percent coarse aggregate that you want, the specific gravity of coarse aggregate, the volume of total aggregate, and the unit weight of water.  
```

**Inputs:** percent of coarse aggregate (%), specific gravity of coarse aggregate, volume of total aggregate (ft^3), and unit weight of water (lb/yd^3) 

**Outputs:** total weight of coarse aggregate (lb)

```python
def calculate_other_aggregate_AA(percent_other_I, sg_other_P, volume_X):
    weight_AA = unit_weight_water * (percent_other_I / 100) * sg_other_P * volume_X
    return weight_AA
# This function is used to find the total weight of the other aggregates, the percent other aggregate that you want, the specific gravity of the other aggregate, the volume of total aggregate, and the unit weight of water.
```

**Inputs:** percent of other aggregate (%), specific gravity of other aggregate, volume of total aggregate (ft^3), and unit weight of water (lb/yd^3)

**Outputs:** total weight of other aggregate (lb)
